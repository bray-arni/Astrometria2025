## Metodo de la transformada inversa: distribucion de Fisher-Tippet


```python
#importamos las librerias necesarias
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Definimos la funcion de distribucion acumulada (CDF) inversa

def inversa_cdf (y):
    x = -np.log(-np.log(y)) 
    return x

#Generamos numeros aleatorios uniformes entre 0 y 1

y = np.random.uniform(0, 1, 10000)

#Aplicamos la transformacion inversa

x = inversa_cdf(y)

#Graficamos el histograma de los datos generados y la funcion de distribucion teorica
plt.hist(x, bins=50, density=True, alpha=0.6, color='g', label='Datos generados')
x_theo = np.linspace(-5, 5, 1000)
pdf_theo = np.exp(-(x_theo + np.exp(-x_theo)))
plt.plot(x_theo, pdf_theo, 'r-', lw=2, label='Distribución Teórica')
plt.xlabel('valores')
plt.ylabel('Densidad de Probabilidad')
plt.title('Generación de Números Aleatorios con Distribución de Fisher-Tippett')
plt.legend()
plt.show()  


print("Media:", np.mean(x))
print("Media Teorica: 0.5772")
```
