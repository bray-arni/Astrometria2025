## Experimento de Buffon

```python
#importamos las librerias necesarias
import numpy as np
```

```python
#definimos la funcion que simula el experimento de Buffon

def buffon_sim(N, l, t):

    cruces = 0  # definimos un contador que nos dira cuantas veces cruza la aguja una raya

    for _ in range(N): # Repetimos el experimento N veces
        x = np.random.uniform(0, t/2)  # Distancia desde el centro de la aguja a la raya más cercana 
        theta = np.random.uniform(0, np.pi/2)  # Ángulo de la aguja con la horizontal


        # imponemos la condición para que la aguja cruce una raya
        if (l/2) * np.sin(theta) > x:
            cruces += 1 #aca le sumamos 1 al contador cada vez que la aguja cruza una raya, es decir, que se cumple la condicion


    # Estimos pi usando la fórmula de Buffon
    pi_est = (2 * l * N) / (t * cruces)
    return pi_est

# Definimos los parametros

N = 1000000  # Número de simulaciones
l = 29  # Longitud de la aguja
t = 30  # Separación entre las rayas (debe ser mayor que l)

# Realizamos la simulación

pi_estimado = buffon_sim(N, l, t)

print(f"Estimación de pi después de {N} lanzamientos: {pi_estimado}")

print(f"Valor real de pi: {np.pi}")
```
