## Metodo de la Transformada Inversa: Distribucion de Poisson

```python
#importamos las librerias necesarias

import numpy as np
import matplotlib.pyplot as plt
```

```python
#definimos la funcion inversa de la cdf correspondiente a la pdf de poisson

def inversa_cdf_poisson(y,lam):
    x = -np.log(1-y)/lam
    return x


#generamos los numeros aleatorios uniformes

np.random.seed(42) #Fijamos semilla para reproducibilidad
N = 25
y = np.random.uniform(0,1,N)

#definimos los parametros necesarios

lam = 5

#aplicamos la funcion inversa de la cdf para obtener los numeros aleatorios con distribucion de poisson

x = inversa_cdf_poisson(y,lam)
x = np.append(x,0) #agregamos el valor 0 correspondiente al tiempo incial
suma = np.cumsum(x) #calculamos la suma acumulada para obtener los tiempos de llegada


plt.plot(np.arange(N+1), suma)
plt.title('Proceso Poisson: Distribución de tiempos')
plt.xlabel('Eventos')
plt.ylabel('Tiempo (horas)')
plt.grid(True)


plt.show()
```
